```python
split_data = {}

for ticker in tickers:
    try:
        actions = yf.Ticker(f"{ticker}.NS").actions
        splits = actions[actions['Stock Splits'] != 0]['Stock Splits']

        split_data[ticker] = splits

        if len(splits) > 0:
            print(f"{ticker}:")
            print(splits)
    except Exception as e:
        print(f"{ticker}: ERROR - {e}")
```

```python
def get_split_factor(ticker, date):
    splits = split_data.get(ticker)

    if splits is None or len(splits) == 0:
        return 1.0

    factor = 1.0

    for split_date, split_ratio in splits.items():
        split_date = split_date.tz_localize(None)

        if date < split_date:
            factor *= split_ratio

    return factor


# Test EICHERMOT
for date in pd.to_datetime([
    '2020-08-20',
    '2020-08-21',
    '2020-08-24',
    '2020-08-25'
]):
    print(date.date(), get_split_factor('EICHERMOT', date))
```

```python
daily_avg_corr = {}

for date in returns_matrix.index:
    window = returns_matrix.loc[:date].tail(60)

    if len(window) < 60:
        continue

    corr = window.corr(min_periods=40)

    # Remove diagonal (stock with itself)
    mask = ~np.eye(len(corr), dtype=bool)

    daily_avg_corr[date] = corr.where(mask).stack().mean()

daily_avg_corr = pd.Series(daily_avg_corr, name='Avg_Correlation')

print(daily_avg_corr.shape)
print(daily_avg_corr.loc['2020-03-16':'2020-03-25'])
```

```python
def classify_event(row):

    stress = row['Market_Stress_Score']
    breadth = row['Anomaly_Breadth']
    neg_ratio = row['Negative_Ratio']
    pos_ratio = row['Positive_Ratio']
    corr_score = row['Correlation_Score']

    if stress >= 40 and breadth >= 0.10:
        if neg_ratio >= 0.75:
            return 'Broad Market Downside Shock'
        elif pos_ratio >= 0.75:
            return 'Broad Market Upside Shock'

    if corr_score >= 80 and breadth < 0.10:
        return 'Market-Wide Synchronization'

    if stress >= 25:
        return 'Elevated Market Stress'

    return 'Normal'
```

```python
market_features['Event_Type'] = market_features.apply(
    classify_event,
    axis=1
)
```

```python
crisis_avg_corr = {}

for date in crisis_returns.index:

    window = (
        df[
            df['Date'] <= date
        ]
        .pivot(
            index='Date',
            columns='CanonicalSymbol',
            values='Return'
        )
        .tail(60)
    )

    if len(window) < 60:
        continue

    corr = window.corr(min_periods=40)

    mask = ~np.eye(len(corr), dtype=bool)

    crisis_avg_corr[date] = (
        corr.where(mask).stack().mean()
    )

crisis_avg_corr = pd.Series(
    crisis_avg_corr,
    name='Avg_Correlation'
)

print(crisis_avg_corr.describe())
```

```python
crisis_rolling_mean = (
    crisis_avg_corr
    .shift(1)
    .rolling(252)
    .mean()
)

crisis_rolling_std = (
    crisis_avg_corr
    .shift(1)
    .rolling(252)
    .std()
)

crisis_corr_z = (
    (crisis_avg_corr - crisis_rolling_mean)
    / crisis_rolling_std
)

crisis_corr_score = crisis_corr_z.clip(lower=0)

print(crisis_corr_score.describe())
```
