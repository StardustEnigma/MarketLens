```python
split_data = {}

for ticker in tickers:
    try:
        actions = yf.Ticker(f"{ticker}.NS").actions
        splits = actions[actions['Stock Splits'] != 0]['Stock Splits']

        split_data[ticker] = splits

        if len(splits) > 0:
            print(f"{ticker}:")
            print(splits)
    except Exception as e:
        print(f"{ticker}: ERROR - {e}")
```

```python
def get_split_factor(ticker, date):
    splits = split_data.get(ticker)

    if splits is None or len(splits) == 0:
        return 1.0

    factor = 1.0

    for split_date, split_ratio in splits.items():
        split_date = split_date.tz_localize(None)

        if date < split_date:
            factor *= split_ratio

    return factor


# Test EICHERMOT
for date in pd.to_datetime([
    '2020-08-20',
    '2020-08-21',
    '2020-08-24',
    '2020-08-25'
]):
    print(date.date(), get_split_factor('EICHERMOT', date))
```

```python
eicher_test = df[df['CanonicalSymbol'] == 'EICHERMOT'].copy()

eicher_test['Split_Factor'] = eicher_test['Date'].apply(
    lambda x: get_split_factor('EICHERMOT', x)
)

eicher_test['Adjusted_Volume'] = (
    eicher_test['Volume'] * eicher_test['Split_Factor']
)

print(
    eicher_test[
        eicher_test['Date'].between('2020-08-20', '2020-08-26')
    ][['Date', 'Volume', 'Split_Factor', 'Adjusted_Volume']]
)
```

```python
eicher_test['Volume_MA_20_Adjusted'] = (
    eicher_test['Adjusted_Volume']
    .rolling(20)
    .mean()
)

eicher_test['Volume_Ratio_Adjusted'] = (
    eicher_test['Adjusted_Volume']
    / eicher_test['Volume_MA_20_Adjusted']
)

print(
    eicher_test[
        eicher_test['Date'].between('2020-08-20', '2020-08-26')
    ][['Date', 'Adjusted_Volume', 'Volume_Ratio_Adjusted']]
)
```
